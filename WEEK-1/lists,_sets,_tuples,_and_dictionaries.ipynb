```python
fruits = ['apple', 'banana', 'cherry', 'apple']
print("Original list:", fruits)
print("First fruit:", fruits[0])

```

```python
fruits.append('orange')
print("fruits after appeding:",fruits)
```

```python
fruits.remove('banana')
print("after removing:",fruits)
print("no of fruits:",len(fruits))
```

```python
dimensions=(10,20,30)
print("tuple:",dimensions)
```

```python
print("second element:",dimensions[1])
```

```python
unique_numbers={1,2,3,2,1}
print("set(no duplicates):",unique_numbers)
```

```python
unique_numbers.add(5)
print("after adding 5:",unique_numbers)
```

```python
unique_numbers.remove(2)
print("removing:",unique_numbers)
```

```python
set_a={1,2,3}
set_b={4,3,5}
print("union:",set_a | set_b)
print("intersection:",set_a & set_b)
print("difference(a-b):",set_a - set_b)
```

```python
student={
    'name':'alice',
    'age': 19,
    'course':'python'
}
print("dictionary:",student)
```

```python
print("student name:",student['name'])
```

```python
student['grade']='A'
print('after adding:',student)

del student['age']
print('after removing:',student)
for key, value in student.items():
    print(f"{key}: {value}")
```

```python
keys = ['id', 'name', 'marks']
values = [101, 'John', 88]
student_info=dict(zip(keys,values))
print('info',student_info)
```
